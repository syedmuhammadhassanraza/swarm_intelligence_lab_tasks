```python
# Import the random module for generating random numbers
import random
# Seed the random number generator for reproducibility
random.seed(80)

# Define the objective function to be minimized
def f(x):
    return (x - 7) ** 2 + 4

# Initialize particle positions randomly within a range (0 to 14)
positions = [random.uniform(0, 14) for _ in range(10)]
# Initialize particle velocities to zero
velocities = [0 for _ in range(10)]
# Initialize personal best positions (pbest_pos) with initial positions
pbest_pos = positions[:]
# Initialize global best position (gbest_pos) as the best among initial pbest_pos
gbest_pos = min(pbest_pos, key=f)
# Set PSO parameters: inertia weight (w), cognitive coefficient (c1), and social coefficient (c2)
w, c1, c2 = 0.6, 1.5, 1.5
# Set the number of iterations for the optimization process
iterations = 40
```

```python
# Main loop for the Particle Swarm Optimization algorithm
for it in range(iterations):
    # Iterate through each particle
    for i in range(len(positions)):
        # Generate two random numbers for velocity update equation
        r1, r2 = random.random(), random.random()
        # Update particle velocity based on its current velocity, personal best, and global best
        velocities[i] = w * velocities[i] + c1 * r1 * (pbest_pos[i] - positions[i]) + c2 * r2 * (gbest_pos - positions[i])
        # Update particle position based on its new velocity
        positions[i] = positions[i] + velocities[i]

        # Apply boundaries to particle positions (confine them within 0 and 14)
        if positions[i] < 0:
            positions[i] = 0
        elif positions[i] > 14:
            positions[i] = 14

        # Update personal best position if the current position is better
        if f(positions[i]) < f(pbest_pos[i]):
            pbest_pos[i] = positions[i]
    # Update global best position after all particles have moved in the current iteration
    gbest_pos = min(pbest_pos, key=f)

# Print the final converged global best position
print("Swarm converged near x =", gbest_pos)
```
