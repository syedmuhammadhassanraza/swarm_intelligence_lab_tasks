```python
import numpy as np
import matplotlib.pyplot as plt
def sphere(x, y):
 return x**2 + y**2
```

```python
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)
X, Y = np.meshgrid(x, y)
Z = sphere(X, Y)
plt.contourf(X, Y, Z, levels=30, cmap='viridis')
plt.colorbar(label='f(x, y)')
plt.title('Sphere Function Landscape')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
def rastrigin(x, y):
 A = 10
 return 2*A + (x**2 - A*np.cos(2*np.pi*x)) \
 + (y**2 - A*np.cos(2*np.pi*y))
x = np.linspace(-5.12, 5.12, 200)
y = np.linspace(-5.12, 5.12, 200)
X, Y = np.meshgrid(x, y)
Z = rastrigin(X, Y)
plt.contourf(X, Y, Z, levels=30, cmap='viridis')
plt.colorbar(label='f(x, y)')
plt.title('Rastrigin Function Landscape')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
def ackley(x, y):
    return (-20*np.exp(-0.2*np.sqrt(0.5*(x**2+y**2)))
            - np.exp(0.5*(np.cos(2*np.pi*x)
            + np.cos(2*np.pi*y)))
            + np.e + 20)
x = np.linspace(-5, 5, 200)
y = np.linspace(-5, 5, 200)
X, Y = np.meshgrid(x, y)
Z = ackley(X, Y)
plt.contourf(X, Y, Z, levels=30, cmap='viridis')
plt.colorbar(label='f(x, y)')
plt.title('Ackley Function Landscape')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

```python
x_ackley = np.linspace(-32.768, 32.768, 200)
y_ackley = np.linspace(-32.768, 32.768, 200)
X_ackley, Y_ackley = np.meshgrid(x_ackley, y_ackley)
Z_ackley = ackley(X_ackley, Y_ackley)

plt.figure(figsize=(7, 5.5))
contour_ackley = plt.contourf(X_ackley, Y_ackley, Z_ackley, levels=30, cmap="viridis")
plt.colorbar(contour_ackley, label="f(x, y)")
plt.title("Ackley Function Landscape (Standard Domain)")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
import random
def random_search(func, bounds, iterations=1000):
    best_point = None
    best_score = float('inf')
    for i in range(iterations):
        x = random.uniform(bounds[0], bounds[1])
        y = random.uniform(bounds[0], bounds[1])
        score = func(x, y)
        if score < best_score:
            best_score = score
            best_point = (x, y)
    return best_point, best_score
```

```python
best_point, best_score = random_search(sphere, (-5, 5))
print('Sphere -> Best point:', best_point,
 'Best score:', best_score)
best_point, best_score = random_search(rastrigin, (-5.12, 5.12))
print('Rastrigin -> Best point:', best_point,
 'Best score:', best_score)
best_point, best_score = random_search(ackley, (-5, 5))
print('Ackley -> Best point:', best_point,
 'Best score:', best_score)
```

**question 1: Which function did Random Search solve best? Why do you think that happened, based on its landscape shape?**

Random Search did best on the Sphere function, getting a best score of 0.00056. Sphere has one smooth bowl shaped minimum with no other dips or bumps, so random points landing anywhere near the center still score well and its easy to stumble onto a good result.

**question 2: Which function did Random Search struggle with the most? What about its landscape made it difficult?**

It struggled most with Rastrigin, ending with a best score of 1.019. Rastrigin is full of small peaks and valleys spread across the space, so most random points fall into one of many local dips instead of the actual global minimum. There are just too many misleading spots.

**question 3: If you increased iterations from 1000 to 10000, do you expect the results to improve? For which function would the improvement matter least, and why?**

Yes, more iterations should help across the board since more points get tested and the chances of hitting a better one go up. But the gain would matter least for Sphere, because it already reached a near zero score with just 1000 tries. Its shape is so simple that extra attempts wont change the result by much.

**question 4: In your own words, what is the main weakness of Random Search as a strategy?**

The biggest problem with Random Search is that it has no memory or direction. It doesn't use past results to guide future guesses, so it keeps trying points all over the place even after finding a decent one. That makes it slow and wasteful, especially on harder problems with more complex shapes.
